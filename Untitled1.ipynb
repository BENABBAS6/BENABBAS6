```python
import requests

def get_movie_details(movie_id: int):
    url = f"https://api.themoviedb.org/3/movie/{movie_id}"
    params = {
        "api_key": TMDB_API_KEY,
        "language": "fr-FR"
    }

    try:
        response = requests.get(url, params=params)

        # Gestion des erreurs (Q1 commune / robustesse)
        if response.status_code == 401:
            print("Erreur 401 : Clé API invalide.")
            return None
        elif response.status_code == 404:
            print(f"Erreur 404 : Film ID {movie_id} introuvable.")
            return None

        response.raise_for_status()
        data = response.json()

        # Extraction des informations d'intérêt
        return {
            "id": data.get("id"),
            "titre": data.get("title"),
            "date_sortie": data.get("release_date"),
            "genres": [g["name"] for g in data.get("genres", [])],
            "popularite": data.get("popularity"),
            "note_moyenne": data.get("vote_average")
        }
    except Exception as e:
        print(f"Erreur technique : {e}")
        return None

# Test pratique sur l'ID 550 (Fight Club)
print(get_movie_details(550))
```

```python
import gzip
import json
from datetime import datetime
import pandas as pd

def telecharger_et_decompresser():
    # Date du jour au format MM_DD_YYYY
    target_date = datetime.now().strftime("%m_%d_%Y")

    url = f"http://files.tmdb.org/p/exports/movie_ids_{target_date}.json.gz"
    gz_filename = f"movie_ids_{target_date}.json.gz"
    json_filename = f"movie_ids_{target_date}.json"

    print(f"Téléchargement de l'export du {target_date}...")
    response = requests.get(url, stream=True)

    if response.status_code != 200:
        print(f"Erreur : Fichier introuvable (Code {response.status_code}). Rappel : les fichiers de plus de 3 mois sont supprimés.")
        return None

    # Enregistrement du fichier .gz
    with open(gz_filename, "wb") as f:
        for chunk in response.iter_content(chunk_size=8192):
            f.write(chunk)

    # Décomposition du fichier .gz
    print("Décompression du fichier...")
    with gzip.open(gz_filename, "rt", encoding="utf-8") as f_in, open(json_filename, "w", encoding="utf-8") as f_out:
        for line in f_in:
            f_out.write(line)

    # Transformation en array JSON / DataFrame
    data_list = []
    with open(json_filename, "r", encoding="utf-8") as f:
        for line in f:
            if line.strip():
                data_list.append(json.loads(line))

    df = pd.DataFrame(data_list)
    print(f"DataFrame créé avec succès : {len(df)} films chargés.")
    return df
```

```python
from sqlalchemy import create_engine

def injecter_dans_sqlite(df):
    if df is None or df.empty:
        print("Aucune donnée à injecter.")
        return

    # Création de la base SQLite locale dans Google Colab
    engine = create_engine("sqlite:///tmdb_movies.db")

    print("Injection des données dans la base SQLite (cela prend quelques secondes)...")
    df.to_sql(
        "tmdb_movie_ids",
        con=engine,
        if_exists="replace",
        index=False,
        chunksize=10000  # Insertion par blocs de 10 000 lignes
    )
    print("Injection terminée avec succès dans la table 'tmdb_movie_ids' !")

# Exécution globale des étapes E et F :
df_movies = telecharger_et_decompresser()
injecter_dans_sqlite(df_movies)
```
