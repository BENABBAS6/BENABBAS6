```python
import requests
import pandas as pd

# ==========================================
# 1. Fonction pour afficher les stats d'un Pokémon
# ==========================================
def get_pokemon_stats(identifier):
    """
    Récupère et affiche les statistiques clés d'un Pokémon par son nom ou son ID.
    """
    url = f"https://pokeapi.co/api/v2/pokemon/{str(identifier).lower()}"
    response = requests.get(url)

    if response.status_code != 200:
        print(f"❌ Erreur : Impossible de trouver le Pokémon '{identifier}' (Code {response.status_code}).")
        return None

    data = response.json()

    # Extraction propre des statistiques (sans retourner le gros JSON brut)
    stats = {}
    for stat_entry in data.get("stats", []):
        stat_name = stat_entry["stat"]["name"]
        stat_value = stat_entry["base_stat"]
        stats[stat_name] = stat_value

    pokemon_info = {
        "nom": data.get("name").capitalize(),
        "id": data.get("id"),
        "poids": data.get("weight"),
        "taille": data.get("height"),
        "hp": stats.get("hp", 0),
        "attack": stats.get("attack", 0),
        "defense": stats.get("defense", 0),
        "special-attack": stats.get("special-attack", 0),
        "special-defense": stats.get("special-defense", 0),
        "speed": stats.get("speed", 0)
    }

    print(f"\n--- Statistiques de {pokemon_info['nom']} (ID: {pokemon_info['id']}) ---")
    print(f"❤️ Points de vie (HP) : {pokemon_info['hp']}")
    print(f"⚔️ Attaque            : {pokemon_info['attack']}")
    print(f"🛡️ Défense            : {pokemon_info['defense']}")
    print(f"🔮 Attaque Spéciale   : {pokemon_info['special-attack']}")
    print(f"🛡️ Défense Spéciale   : {pokemon_info['special-defense']}")
    print(f"⚡ Vitesse            : {pokemon_info['speed']}")

    return pokemon_info


# ==========================================
# 2. Fonction pour comparer deux Pokémon
# ==========================================
def compare_two_pokemons(poke1_name, poke2_name):
    """
    Compare les statistiques de deux Pokémon et détermine le plus fort en HP et en Attaque.
    """
    print(f"\n--- Comparaison : {poke1_name} vs {poke2_name} ---")
    p1 = get_pokemon_stats(poke1_name)
    p2 = get_pokemon_stats(poke2_name)

    if not p1 or not p2:
        return

    print("\n[Résultat de la comparaison]")
    # Comparaison HP
    if p1["hp"] > p2["hp"]:
        print(f"🏆 HP : {p1['nom']} est plus résistant ({p1['hp']} vs {p2['hp']})")
    elif p2["hp"] > p1["hp"]:
        print(f"🏆 HP : {p2['nom']} est plus résistant ({p2['hp']} vs {p1['hp']})")
    else:
        print("🤝 HP : Égalité parfaite en points de vie !")

    # Comparaison Attaque
    if p1["attack"] > p2["attack"]:
        print(f"🏆 Attaque : {p1['nom']} tape plus fort ({p1['attack']} vs {p2['attack']})")
    elif p2["attack"] > p1["attack"]:
        print(f"🏆 Attaque : {p2['nom']} tape plus fort ({p2['attack']} vs {p1['attack']})")
    else:
        print("🤝 Attaque : Égalité parfaite en attaque !")


# ==========================================
# 3. Fonction d'analyse par Type (ex: fire, dragon)
# ==========================================
def analyze_pokemon_type(type_name):
    """
    Calcule le nombre de Pokémon d'un type donné et la moyenne de leurs points de vie (HP).
    """
    url = f"https://pokeapi.co/api/v2/type/{type_name.lower()}"
    response = requests.get(url)

    if response.status_code != 200:
        print(f"❌ Erreur : Type '{type_name}' introuvable.")
        return

    data = response.json()
    pokemon_list = data.get("pokemon", [])

    total_hp = 0
    count = 0

    print(f"\n--- Analyse du type : {type_name.upper()} ---")
    print(f"Nombre total de Pokémon répertoriés pour ce type : {len(pokemon_list)}")
    print("Calcul de la moyenne des points de vie en cours (veuillez patienter)...")

    # Pour éviter de surcharger l'API, on analyse un échantillon ou la liste complète
    # (Ici on prend les 20 premiers pour aller vite, ou toute la liste si besoin)
    for p in pokemon_list[:20]:
        p_url = p["pokemon"]["url"]
        p_data = requests.get(p_url).json()
        for stat in p_data.get("stats", []):
            if stat["stat"]["name"] == "hp":
                total_hp += stat["base_stat"]
                count += 1

    if count > 0:
        avg_hp = total_hp / count
        print(f"📊 Moyenne des HP (sur un échantillon de {count} Pokémon) : {avg_hp:.2f}")


# ==========================================
# 4. Simulation d'un combat sur 5 tours
# ==========================================
def simulate_battle(poke1_name, poke2_name):
    """
    Simule un combat entre deux Pokémon sur 5 tours en fonction de l'attaque et de la défense.
    Le Pokémon qui inflige le plus de dégâts totaux en 5 tours gagne.
    """
    print(f"\n⚔️ --- DÉBUT DU COMBAT : {poke1_name.upper()} vs {poke2_name.upper()} --- ⚔️")
    p1 = get_pokemon_stats(poke1_name)
    p2 = get_pokemon_stats(poke2_name)

    if not p1 or not p2:
        return

    # Calcul simple des dégâts par tour (Attaque - Défense adverse, avec un minimum de 10 dégâts)
    dmg_p1_to_p2 = max(10, p1["attack"] - (p2["defense"] // 2))
    dmg_p2_to_p1 = max(10, p2["attack"] - (p1["defense"] // 2))

    total_dmg_p1 = 0
    total_dmg_p2 = 0

    for turn in range(1, 6):
        print(f"\n--- Tour {turn} ---")
        print(f"🔥 {p1['nom']} inflige {dmg_p1_to_p2} dégâts à {p2['nom']}.")
        print(f"🔥 {p2['nom']} inflige {dmg_p2_to_p1} dégâts à {p1['nom']}.")
        total_dmg_p1 += dmg_p1_to_p2
        total_dmg_p2 += dmg_p2_to_p1

    print(f"\n==============================")
    print(f"📈 RÉSULTATS FINAUX (après 5 tours) :")
    print(f"- Dégâts totaux infligés par {p1['nom']} : {total_dmg_p1}")
    print(f"- Dégâts totaux infligés par {p2['nom']} : {total_dmg_p2}")

    if total_dmg_p1 > total_dmg_p2:
        print(f"🏆 VAINQUEUR : {p1['nom']} remporte le combat !")
    elif total_dmg_p2 > total_dmg_p1:
        print(f"🏆 VAINQUEUR : {p2['nom']} remporte le combat !")
    else:
        print("🤝 Match nul ! C'est une égalité parfaite.")
```

```python
get_pokemon_stats("charizard")
```

```python
compare_two_pokemons("pikachu", "mewtwo")
```

```python
analyze_pokemon_type("fire")
```

```python
simulate_battle("charizard", "blastoise")
```
